````python
import math

# 16 people: exposure to the factor (1/0) and disease (1/0)
exposed = [1] * 8 + [0] * 8
ill = [1] * 6 + [0] * 2 + [1] * 2 + [0] * 6
n = len(exposed)

def sigmoid(z):
    return 1 / (1 + math.exp(-z))

b0, b1 = 0.0, 0.0                # intercept and coefficient
ETA = 1.0                        # learning rate
for step in range(5000):  # gradient descent (cross-entropy)
    g0 = g1 = 0.0
    for x, y in zip(exposed, ill):
        error = sigmoid(b0 + b1 * x) - y
        g0 += error / n
        g1 += error * x / n
    b0 -= ETA * g0
    b1 -= ETA * g1

print(f"intercept {b0:.2f}, coefficient {b1:.2f}")
print(f"odds ratio {math.exp(b1):.1f}")
print(f"P(disease | not exposed) = {sigmoid(b0):.2f}")
print(f"P(disease | exposed) = {sigmoid(b0 + b1):.2f}")
````
