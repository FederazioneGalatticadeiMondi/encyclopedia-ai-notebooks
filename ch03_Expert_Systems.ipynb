````python
# Knowledge base: (id, premises, conclusion,
# certainty factor of the rule)
RULES = [
    ("R1", ["starter_not_turning", "dim_lights"], "flat_battery", 0.8),
    ("R2", ["starter_not_turning", "normal_lights"], "faulty_starter", 0.6),
    ("R3", ["starter_turning", "smell_of_petrol"], "flooded_engine", 0.7),
    ("R4", ["starter_turning", "fuel_warning_light"], "empty_tank", 0.9),
    ("R5", ["old_battery"], "flat_battery", 0.4),
]

def combine(cf1, cf2):
    """Combine two pieces of positive evidence (MYCIN model)."""
    return cf1 + cf2 * (1 - cf1)

def consult(observations, threshold=0.2):
    facts = dict(observations)      # fact -> degree of certainty
    explanations = {}
    for name, premises, conclusion, cf_rule in RULES:
        cf_premises = min(facts.get(p, 0.0) for p in premises)
        if cf_premises <= threshold:
            continue                # premises too uncertain: rule ignored
        cf = cf_rule * cf_premises
        if conclusion in facts:
            facts[conclusion] = combine(facts[conclusion], cf)
        else:
            facts[conclusion] = cf
        explanations.setdefault(conclusion, []).append((name, premises, cf))
    return facts, explanations

observations = {"starter_not_turning": 1.0, "dim_lights": 0.8,
                "old_battery": 0.9}

facts, explanations = consult(observations)
for hypothesis, rules in explanations.items():
    print(f"{hypothesis}: certainty {facts[hypothesis]:.2f}")
    for name, premises, cf in rules:
        print(f"  because {name}: {' and '.join(premises)} (contribution {cf:.2f})")
````
