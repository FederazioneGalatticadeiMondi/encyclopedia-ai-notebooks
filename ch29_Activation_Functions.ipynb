````python
import math

def sigmoid(z):
    return 1 / (1 + math.exp(-z))

def relu(z):
    return max(0.0, z)

def leaky_relu(z, alpha=0.01):
    return z if z > 0 else alpha * z

def gelu(z):
    """z times the probability that a standard normal is less than z."""
    return z * 0.5 * (1 + math.erf(z / math.sqrt(2)))

FUNCTIONS = [("sigmoid", sigmoid), ("tanh", math.tanh), ("ReLU", relu),
             ("Leaky ReLU", leaky_relu), ("GELU", gelu)]
POINTS = [-2.0, 0.0, 2.0]

print(f"{'z':<10}" + "".join(f"{z:7.3f}" for z in POINTS))
for name, f in FUNCTIONS:
    print(f"{name:<10}" + "".join(f"{f(z):7.3f}" for z in POINTS))

# Vanishing gradient: product of the derivatives over 10 layers
max_derivative = sigmoid(0) * (1 - sigmoid(0))      # 0.25
print(f"maximum derivative of the sigmoid: {max_derivative:.3f}")
print(f"product over 10 layers: sigmoid {max_derivative ** 10:.2e}, "
      f"ReLU {1.0 ** 10}")
````
