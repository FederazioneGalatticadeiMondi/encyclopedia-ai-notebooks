````python
# Max pooling, average pooling and global average pooling on a 4x4 map
feature_map = [[1, 3, 2, 0],
               [4, 6, 1, 1],
               [0, 2, 5, 7],
               [1, 1, 3, 2]]

def pooling(m, function, k=2, stride=2):
    """Apply the summary function to k x k windows with the given stride."""
    side = (len(m) - k) // stride + 1
    return [[function([m[stride * i + u][stride * j + v]
                       for u in range(k) for v in range(k)])
             for j in range(side)]
            for i in range(side)]

def mean(values):
    return sum(values) / len(values)

maxima = pooling(feature_map, max)
averages = pooling(feature_map, mean)
print("max pooling 2x2:")
for row in maxima:
    print(" ".join(str(v) for v in row))
print("average pooling 2x2:")
for row in averages:
    print(" ".join(f"{v:.2f}" for v in row))
print("global average pooling:", sum(sum(r) for r in feature_map) / 16)

# Moving the 6 within the same window does not change the maximum
shifted = [row[:] for row in feature_map]
shifted[0][0], shifted[1][1] = 6, 1
print("max pooling unchanged after the shift:",
      pooling(shifted, max) == maxima)
````
