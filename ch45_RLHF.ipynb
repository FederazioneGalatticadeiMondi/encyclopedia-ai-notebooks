````python
import math

def sigma(z):
    return 1 / (1 + math.exp(-z))

def preference_loss(r_chosen, r_rejected):
    """Bradley-Terry model: P(chosen preferred) and loss."""
    p = sigma(r_chosen - r_rejected)
    return p, -math.log(p)

# 1. Reward model
p, L = preference_loss(2.0, 0.5)
print(f"correct preference: P = {p:.4f}, loss = {L:.4f}")
p, L = preference_loss(0.5, 2.0)
print(f"reversed preference: P = {p:.4f}, loss = {L:.4f}")

# 2. DPO: log-probabilities under the policy and the reference
beta = 0.1
lp_chosen, ref_chosen = -10.0, -12.0
lp_rejected, ref_rejected = -15.0, -14.0
margin = beta * ((lp_chosen - ref_chosen)
                 - (lp_rejected - ref_rejected))
p = sigma(margin)
print(f"DPO: margin = {margin:.2f}, P = {p:.4f}")
print(f"DPO: loss = {-math.log(p):.4f}")
````
