````python
# Perceptron learning rule on AND and XOR
INPUTS = [(0, 0), (0, 1), (1, 0), (1, 1)]
AND = [0, 0, 0, 1]
XOR = [0, 1, 1, 0]

def predict(w, b, x):
    return 1 if w[0] * x[0] + w[1] * x[1] + b > 0 else 0

def epoch(w, b, labels):
    """One pass over the examples; returns weights, bias and number of mistakes."""
    mistakes = 0
    for x, y in zip(INPUTS, labels):
        delta = y - predict(w, b, x)        # 0 if right, +1 or -1 if wrong
        if delta != 0:
            mistakes += 1
            w = [w[0] + delta * x[0], w[1] + delta * x[1]]
            b = b + delta
    return w, b, mistakes

print("AND")
w, b = [0, 0], 0
for n in range(1, 21):
    w, b, mistakes = epoch(w, b, AND)
    print(f"epoch {n}: mistakes {mistakes}, w = {w}, b = {b}")
    if mistakes == 0:
        print(f"learned in {n} epochs")
        break

print("XOR")
w, b = [0, 0], 0
mistakes_per_epoch = []
for n in range(20):
    w, b, mistakes = epoch(w, b, XOR)
    mistakes_per_epoch.append(mistakes)
if min(mistakes_per_epoch) > 0:
    print("still mistakes after 20 epochs: not linearly separable")
````
