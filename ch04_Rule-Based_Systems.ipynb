````python
# Each rule: name, priority, condition (IF) and action (THEN)
RULES = [
    {"name": "premium_customer", "priority": 10,
     "if": lambda m: "category" not in m and m["total"] >= 500,
     "then": lambda m: m.update(category="premium")},
    {"name": "standard_customer", "priority": 1,
     "if": lambda m: "category" not in m,
     "then": lambda m: m.update(category="standard")},
    {"name": "premium_discount", "priority": 5,
     "if": lambda m: m.get("category") == "premium" and "discount" not in m,
     "then": lambda m: m.update(discount=0.10)},
    {"name": "standard_discount", "priority": 5,
     "if": lambda m: m.get("category") == "standard" and "discount" not in m
                     and m["total"] >= 100,
     "then": lambda m: m.update(discount=0.05)},
    {"name": "no_discount", "priority": 0,
     "if": lambda m: "category" in m and "discount" not in m,
     "then": lambda m: m.update(discount=0.0)},
    {"name": "free_shipping", "priority": 3,
     "if": lambda m: "discount" in m and "shipping" not in m
                     and m["total"] * (1 - m["discount"]) >= 300,
     "then": lambda m: m.update(shipping=0.0)},
    {"name": "standard_shipping", "priority": 2,
     "if": lambda m: "discount" in m and "shipping" not in m,
     "then": lambda m: m.update(shipping=9.90)},
]

def run(memory, rules, max_cycles=100):
    """Recognise-act cycle; conflicts resolved by priority."""
    for cycle in range(1, max_cycles + 1):
        can_fire = [r for r in rules if r["if"](memory)]       # recognise
        if not can_fire:
            break
        chosen = max(can_fire, key=lambda r: r["priority"])    # conflicts
        chosen["then"](memory)                                 # act
        names = [r["name"] for r in can_fire]
        print(f"cycle {cycle}: can fire {names} -> fired {chosen['name']}")
    return memory

order = {"total": 600}
print("final memory:", run(order, RULES))
````

## Expected output

If everything works, you should see:

```

cycle 2: can fire ['premium_discount', 'no_discount'] -> fired premium_discount
cycle 3: can fire ['free_shipping', 'standard_shipping'] -> fired free_shipping
final memory: {'total': 600, 'category': 'premium', 'discount': 0.1, 'shipping': 0.0}
```
