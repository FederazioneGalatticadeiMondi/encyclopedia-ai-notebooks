````python
# 1. A few-shot input text for in-context learning
instruction = "Translate English to French."
examples = [("cheese", "fromage"), ("house", "maison"), ("book", "livre")]
new_case = "sea"
lines = [instruction] + [f"{a} -> {b}" for a, b in examples] + [f"{new_case} ->"]
print("\n".join(lines))

# 2. Scale estimates for a model with L layers and vectors of dimension d
L, d = 96, 12288
estimated_parameters = 12 * L * d ** 2
N, D = 175e9, 300e9                      # stated parameters and tokens
compute = 6 * N * D
chinchilla_tokens = 20 * N
print(f"estimated parameters: {estimated_parameters:.2e}")
print(f"training compute: {compute:.2e} operations")
print(f"tokens suggested by Chinchilla: {chinchilla_tokens:.2e}")
````
