````python
import math

beta = [0.1, 0.2, 0.3, 0.4]   # noise schedule
x0, eps = 2.0, 0.5            # data and fixed noise

# Forward process with the closed form
alpha_bar = 1.0
for t, b in enumerate(beta, start=1):
    alpha_bar *= 1 - b
    signal = math.sqrt(alpha_bar) * x0
    noise = math.sqrt(1 - alpha_bar) * eps
    xt = signal + noise
    print(f"t={t}: alpha_bar={alpha_bar:.4f}, x_t={xt:.4f}")

# Estimate of x0 from the predicted noise
s = math.sqrt(alpha_bar)
r = math.sqrt(1 - alpha_bar)
print(f"x0 with exact noise: {(xt - r * eps) / s:.4f}")
eps_predicted = 0.4
x0_estimated = (xt - r * eps_predicted) / s
print(f"x0 with predicted noise: {x0_estimated:.4f}")
print(f"loss: {(eps - eps_predicted) ** 2:.4f}")
````
