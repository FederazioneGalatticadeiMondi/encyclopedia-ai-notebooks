````python
# Byte-pair encoding (BPE) on a small corpus with frequencies
corpus = {("t", "a", "n"): 4, ("t", "a", "n", "k"): 3,
          ("a", "n", "t"): 2}

def count_pairs(corpus):
    counts = {}
    for word, frequency in corpus.items():
        for a, b in zip(word, word[1:]):
            counts[(a, b)] = counts.get((a, b), 0) + frequency
    return counts

def merge(corpus, pair):
    new = {}
    for word, frequency in corpus.items():
        symbols, i = [], 0
        while i < len(word):
            if i + 1 < len(word) and (word[i], word[i + 1]) == pair:
                symbols.append(word[i] + word[i + 1])
                i += 2
            else:
                symbols.append(word[i])
                i += 1
        new[tuple(symbols)] = frequency
    return new

for step in range(1, 4):
    counts = count_pairs(corpus)
    best = max(counts, key=counts.get)
    print(f"merge {step}: {best[0]} + {best[1]} -> "
          f"{''.join(best)} (frequency {counts[best]})")
    corpus = merge(corpus, best)

for word in corpus:
    print("".join(word), "->", " ".join(word))
````
