````python
# Backpropagation by hand on a tiny network: x -> w1 -> ReLU -> w2 -> y
x, t = 2, 1                  # input and target
w1, w2 = 0.5, 3.0            # initial weights

def forward(w1, w2):
    z = w1 * x
    h = max(0.0, z)          # ReLU
    y = w2 * h
    loss = (y - t) ** 2 / 2
    return z, h, y, loss

z, h, y, J = forward(w1, w2)
print(f"forward: z = {z}, h = {h}, y = {y}, loss = {J}")

# Backward pass with the chain rule
dJ_dy = y - t
dJ_dw2 = dJ_dy * h
dJ_dh = dJ_dy * w2
dJ_dz = dJ_dh * (1.0 if z > 0 else 0.0)    # derivative of ReLU
dJ_dw1 = dJ_dz * x
print(f"backpropagation: dJ/dw1 = {dJ_dw1}, dJ/dw2 = {dJ_dw2}")

# Check with central finite differences
eps = 1e-4
num_w1 = (forward(w1 + eps, w2)[3] - forward(w1 - eps, w2)[3]) / (2 * eps)
num_w2 = (forward(w1, w2 + eps)[3] - forward(w1, w2 - eps)[3]) / (2 * eps)
print(f"finite differences: dJ/dw1 = {num_w1:.4f}, dJ/dw2 = {num_w2:.4f}")

# One gradient descent step
eta = 0.01
new = forward(w1 - eta * dJ_dw1, w2 - eta * dJ_dw2)[3]
print(f"loss after one step (eta = {eta}): {new:.3f}")
````
