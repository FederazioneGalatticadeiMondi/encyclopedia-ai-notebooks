````python
import math

def sigmoid(v):
    return 1 / (1 + math.exp(-v))

# 1. One GRU step with a scalar state (z = part of the old state kept)
h_prev = 1.0
z = sigmoid(2.0)                         # update gate
r = sigmoid(0.0)                         # reset gate
candidate = math.tanh(0.0 + 1.0 * (r * h_prev))   # zero input, weight 1
h = z * h_prev + (1 - z) * candidate     # interpolation between old and new
print(f"GRU step: z = {z:.3f}, r = {r:.3f}, candidate = {candidate:.3f}, "
      f"h = {h:.3f}")

# 2. Parameters: three transformations for the GRU, four for the LSTM
d, n = 100, 256
per_transformation = n * d + n * n + n
print(f"parameters with d = {d}, n = {n}: GRU {3 * per_transformation}, "
      f"LSTM {4 * per_transformation}")
````
