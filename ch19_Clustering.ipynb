````python
# Agglomerative hierarchical clustering with single linkage on six values
values = [0, 1, 3, 10, 14, 30]

def distance(a, b):
    """Single linkage: distance between the two closest elements."""
    return min(abs(x - y) for x in a for y in b)

groups = [[v] for v in values]          # at the start each value is a group
cut = {}  # partitions saved by number of groups
while len(groups) > 1:
    i, j = min(((i, j) for i in range(len(groups))
                for j in range(i + 1, len(groups))),
               key=lambda c: distance(groups[c[0]], groups[c[1]]))
    d = distance(groups[i], groups[j])
    groups[i] = sorted(groups[i] + groups[j])
    del groups[j]
    print(f"merge at distance {d}: {groups[i]}")
    cut[len(groups)] = [list(g) for g in groups]

print("3 groups:", " ".join(str(g) for g in cut[3]))
````
