````python
def metrics(tp, fp, fn, tn):
    """Base and error measures for one group."""
    n = tp + fp + fn + tn
    return {
        "base": (tp + fn) / n,
        "pos": (tp + fp) / n,
        "PPV": tp / (tp + fp),
        "FPR": fp / (fp + tn),
        "FNR": fn / (fn + tp),
    }

# TP, FP, FN, TN for each group
groups = {"A": (45, 15, 5, 35), "B": (15, 5, 10, 70)}
results = {g: metrics(*c) for g, c in groups.items()}
for g, m in results.items():
    values = ", ".join(f"{k} {v:.2f}" for k, v in m.items())
    print(f"group {g}: {values}")

ratio = results["A"]["FPR"] / results["B"]["FPR"]
print(f"ratio of FPRs: {ratio:.2f}")
````
