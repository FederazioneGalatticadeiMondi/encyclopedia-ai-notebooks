````python
from math import comb, exp

# 1. Expected fraction of examples never chosen in a bootstrap sample
print("examples left out of the bootstrap sample:")
for n in (10, 100, 1000):
    print(f"  n = {n}: {(1 - 1 / n) ** n:.3f}")
print(f"  limit 1/e: {exp(-1):.3f}")

# 2. Accuracy of the majority vote of independent trees
def majority_vote(trees, p):
    """Probability that at least half plus one of the trees are right."""
    minimum = trees // 2 + 1
    return sum(comb(trees, k) * p ** k * (1 - p) ** (trees - k)
               for k in range(minimum, trees + 1))

print("majority vote (independent trees, accuracy 0.70):")
for trees in (1, 5, 11):
    name = "tree" if trees == 1 else "trees"
    print(f"  {trees} {name}: {majority_vote(trees, 0.7):.3f}")
````
