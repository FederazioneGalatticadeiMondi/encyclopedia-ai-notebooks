````python
# Linear recurrence h_t = w * h_{t-1} + x_t with a single initial impulse
inputs = [1, 0, 0, 0, 0]

def states(w, sequence):
    h, result = 0.0, []
    for x in sequence:
        h = w * h + x
        result.append(h)
    return result

for w in (0.5, 1.5):
    values = " ".join(f"{h:.4f}" for h in states(w, inputs))
    print(f"w = {w}: h = {values}")

# Effect of the first input on the state after T steps: d h_T / d x_1 = w^(T-1)
T = 20
print(f"effect of the first input after {T} steps:")
for w in (0.5, 1.5):
    print(f"  w = {w} -> {w ** (T - 1):.2e}")
````
