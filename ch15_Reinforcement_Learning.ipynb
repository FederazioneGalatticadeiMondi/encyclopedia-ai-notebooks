````python
# 5-square corridor: start at 0, square 4 is the goal (+1)
GOAL = 4
ACTIONS = ("left", "right")
GAMMA, ALPHA = 0.9, 1.0          # discount and learning rate

def step(state, action):
    """Deterministic environment: returns (new state, reward, done)."""
    new = max(state - 1, 0) if action == "left" else state + 1
    if new == GOAL:
        return new, 1.0, True
    return new, 0.0, False

Q = {(s, a): 0.0 for s in range(GOAL) for a in ACTIONS}

for round_ in range(20):         # systematic exploration: every pair
    for s in range(GOAL):
        for a in ACTIONS:
            s2, r, done = step(s, a)
            future = 0.0 if done else max(Q[(s2, b)] for b in ACTIONS)
            Q[(s, a)] += ALPHA * (r + GAMMA * future - Q[(s, a)])

def best(s):
    return max(ACTIONS, key=lambda a: Q[(s, a)])

for s in range(GOAL):
    print(f"state {s}: left {Q[(s, 'left')]:.3f}  "
          f"right {Q[(s, 'right')]:.3f}  -> {best(s)}")

s, path, done = 0, [0], False
while not done:                  # the learned policy, without exploration
    s, r, done = step(s, best(s))
    path.append(s)
print("path:", " -> ".join(map(str, path)), f"(reward {r:.0f})")
````
