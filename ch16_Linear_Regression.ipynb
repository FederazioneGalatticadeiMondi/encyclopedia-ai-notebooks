````python
# Floor area (m²) and price (thousands of euros) of five flats
areas = [50, 60, 70, 80, 90]
prices = [150, 170, 200, 210, 240]
n = len(areas)

mean_x = sum(areas) / n
mean_y = sum(prices) / n

# Least squares with one variable: slope and intercept
sxy = sum((x - mean_x) * (y - mean_y) for x, y in zip(areas, prices))
sxx = sum((x - mean_x) ** 2 for x in areas)
slope = sxy / sxx
intercept = mean_y - slope * mean_x

def predict(x):
    return intercept + slope * x

# Goodness of fit on the data used for estimation
residuals = [y - predict(x) for x, y in zip(areas, prices)]
rss = sum(r ** 2 for r in residuals)
tss = sum((y - mean_y) ** 2 for y in prices)
r2 = 1 - rss / tss
rmse = (rss / n) ** 0.5

print(f"slope {slope:.2f}, intercept {intercept:.2f}")
print(f"R^2 = {r2:.3f}, RMSE = {rmse:.2f}")
print(f"prediction for 100 m2: {predict(100):.1f} thousand euros")
````
