````python
import math

def dimensions(rw, rh, pixels=1024 * 1024, multiple=64):
    """Sides with aspect ratio rw:rh and about `pixels` pixels in total."""
    w = math.sqrt(pixels * rw / rh)
    h = math.sqrt(pixels * rh / rw)
    w = round(w / multiple) * multiple
    h = round(h / multiple) * multiple
    return w, h

for rw, rh in [(1, 1), (16, 9), (2, 3)]:
    w, h = dimensions(rw, rh)
    print(f"{rw}:{rh} -> {w} x {h}")

# Grids of 4 images and 4 variations of the favourite
requests, per_grid = 3, 4
total = requests * per_grid * 2
print(f"images generated: {total}")
````
