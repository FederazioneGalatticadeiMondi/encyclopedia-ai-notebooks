````python
import math

# 1. Softmax with temperature over the scores of three candidate tokens
scores = [2.0, 1.0, 0.0]

def softmax(values, temperature):
    exponentials = [math.exp(v / temperature) for v in values]
    total = sum(exponentials)
    return [e / total for e in exponentials]

for t in (1.0, 0.5, 2.0):
    probabilities = softmax(scores, t)
    print(f"temperature {t}:", " ".join(f"{p:.3f}" for p in probabilities))

# 2. Greedy generation with a next-word table
next_words = {
    "the": {"cat": 0.6, "dog": 0.4},
    "cat": {"sleeps": 0.7, "eats": 0.3},
    "dog": {"barks": 0.8, "sleeps": 0.2},
    "sleeps": {".": 1.0}, "eats": {".": 1.0}, "barks": {".": 1.0},
}

word, text, text_probability = "the", ["the"], 1.0
while word != ".":
    candidates = next_words[word]
    word = max(candidates, key=candidates.get)   # the most probable
    text_probability *= candidates[word]
    text.append(word)
print("greedy generation:", " ".join(text),
      f"(probability {text_probability:.2f})")
````
