````python
import math

# 1. Batch normalisation of one feature in a batch of 4 examples
batch = [2, 4, 6, 8]
mean = sum(batch) / len(batch)
variance = sum((x - mean) ** 2 for x in batch) / len(batch)
eps = 1e-5
normalised = [(x - mean) / math.sqrt(variance + eps) for x in batch]
gamma, beta = 2, 1
output = [gamma * xn + beta for xn in normalised]
print(f"batch {batch}: mean {mean}, deviation {math.sqrt(variance):.3f}")
print("normalised:", " ".join(f"{v:.3f}" for v in normalised))
print(f"with gamma = {gamma}, beta = {beta}:",
      " ".join(f"{v:.3f}" for v in output))

# 2. Learning rate: linear warm-up, then cosine
def rate(t, maximum=0.1, warmup=2, total=10):
    if t < warmup:
        return maximum * (t + 1) / warmup
    progress = (t - warmup) / (total - warmup)
    return 0.5 * maximum * (1 + math.cos(math.pi * progress))

print("rate:", ", ".join(f"t={t} {rate(t):.3f}" for t in range(0, 11, 2)))
````
