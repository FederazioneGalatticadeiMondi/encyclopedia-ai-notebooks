````python
# Forward pass of a 2-2-1 MLP with ReLU that computes XOR
W1 = [[1, 1], [1, 1]]         # hidden weights (one row per neuron)
b1 = [0, -1]                  # hidden-layer biases
W2 = [1, -2]                  # output-layer weights
b2 = 0                        # output-layer bias

def relu(z):
    return max(0, z)

def layer(W, b, inputs):
    """z = W · inputs + b, then ReLU component by component."""
    return [relu(sum(w * x for w, x in zip(row, inputs)) + bias)
            for row, bias in zip(W, b)]

def mlp(x):
    h = layer(W1, b1, x)
    y = sum(w * hi for w, hi in zip(W2, h)) + b2   # linear output
    return h, y

for x in [(0, 0), (0, 1), (1, 0), (1, 1)]:
    h, y = mlp(x)
    print(f"x = {x} -> h = {tuple(h)} -> y = {y}")

def parameters(dimensions):
    return sum(n * m + m for n, m in zip(dimensions, dimensions[1:]))

print("parameters of the 784-256-128-10 network:", parameters([784, 256, 128, 10]))
````
