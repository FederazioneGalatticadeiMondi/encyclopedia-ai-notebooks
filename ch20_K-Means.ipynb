````python
# Lloyd's algorithm (k-means) on six values, with k = 2
points = [1, 2, 3, 8, 9, 10]
centres = [1.0, 4.0]                    # deliberately poor initialisation

for iteration in range(1, 100):
    # Assignment: each point to the nearest centre
    groups = [[] for _ in centres]
    for x in points:
        j = min(range(len(centres)), key=lambda j: abs(x - centres[j]))
        groups[j].append(x)
    # Update: each centre becomes the mean of its group
    new = [sum(g) / len(g) for g in groups]
    w = sum((x - new[j]) ** 2 for j, g in enumerate(groups) for x in g)
    print(f"iteration {iteration}: centres {new}, W = {w:.2f}")
    if new == centres:
        print(f"converged at iteration {iteration}")
        break
    centres = new
````
