````python
# Feature values with their true labels (used only for evaluation)
DATA = {10: "A", 16: "A", 22: "A", 28: "A", 34: "A", 40: "A", 46: "A",
        60: "B", 66: "B", 72: "B", 78: "B", 84: "B", 90: "B"}
LABELLED = {10: "A", 60: "B"}               # only two known labels

def boundary(labelled):
    """Centroid classifier: midpoint between the class means."""
    means = {}
    for cls in ("A", "B"):
        values = [x for x, y in labelled.items() if y == cls]
        means[cls] = sum(values) / len(values)
    return (means["A"] + means["B"]) / 2

def accuracy(threshold):
    """Accuracy on the examples that were unlabelled at the start."""
    to_evaluate = [x for x in DATA if x not in LABELLED]
    correct = sum(("A" if x < threshold else "B") == DATA[x]
                  for x in to_evaluate)
    return correct / len(to_evaluate)

def self_training(labelled, points):
    """Label one point at a time: the one closest to an already-labelled point."""
    labelled = dict(labelled)
    remaining = [x for x in points if x not in labelled]
    while remaining:
        x, neighbour = min(((x, v) for x in remaining for v in labelled),
                           key=lambda pair: abs(pair[0] - pair[1]))
        labelled[x] = labelled[neighbour]
        remaining.remove(x)
    return labelled

initial_threshold = boundary(LABELLED)
print(f"labelled only: boundary {initial_threshold:.1f}, "
      f"accuracy {accuracy(initial_threshold):.2f}")
propagated = self_training(LABELLED, DATA)
final_threshold = boundary(propagated)
print(f"self-training: boundary {final_threshold:.1f}, "
      f"accuracy {accuracy(final_threshold):.2f}")
````
