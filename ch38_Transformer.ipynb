````python
import math

# 1. Sinusoidal position encodings with vectors of dimension d = 4
d = 4

def position_encoding(pos):
    vector = []
    for i in range(0, d, 2):
        angle = pos / 10000 ** (i / d)
        vector += [math.sin(angle), math.cos(angle)]
    return vector

for pos in range(3):
    values = " ".join(f"{v:.4f}" for v in position_encoding(pos))
    print(f"position {pos}: {values}")

# 2. Parameters of a transformer block with d = 512 and feed-forward 4d
d_model, d_ff = 512, 2048
attention = 4 * d_model * d_model + 4 * d_model
feed_forward = d_model * d_ff + d_ff + d_ff * d_model + d_model
normalisations = 2 * 2 * d_model
total = attention + feed_forward + normalisations
print(f"parameters of one block (d = {d_model}): {total}")
print(f"  attention {attention}, feed-forward {feed_forward}, "
      f"normalisations {normalisations}")
````
