````python
# 1. Functional margin and support vectors for the hyperplane x1 + x2 = 3
points = [((2, 2), +1), ((3, 3), +1), ((4, 2), +1),
          ((0, 0), -1), ((1, 1), -1), ((0, 2), -1)]
w, b = (1, 1), -3

for x, y in points:
    margin = y * (w[0] * x[0] + w[1] * x[1] + b)
    note = "  <- support vector" if margin == 1 else ""
    print(f"{x} class {y:+d}: functional margin {margin}{note}")

norm_w = (w[0] ** 2 + w[1] ** 2) ** 0.5
print(f"margin width: {2 / norm_w:.3f}")

# 2. Kernel trick: (x·z + 1)^2 versus the dot product in 6 dimensions
def phi(v):
    r2 = 2 ** 0.5
    return [v[0] ** 2, v[1] ** 2, r2 * v[0] * v[1], r2 * v[0], r2 * v[1], 1]

x, z = (1, 2), (3, 1)
kernel = (x[0] * z[0] + x[1] * z[1] + 1) ** 2
explicit = sum(a * c for a, c in zip(phi(x), phi(z)))
print(f"kernel (x·z + 1)^2: {kernel:.1f}")
print(f"dot product in 6 dimensions: {explicit:.1f}")
````
