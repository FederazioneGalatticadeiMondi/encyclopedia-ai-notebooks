````python
# 1. Compression in latent space
H, W, C = 512, 512, 3        # colour image
f, c_lat = 8, 4              # reduction factor and channels
pixels = H * W * C
latent = (H // f) * (W // f) * c_lat
print(f"image values: {pixels}, latent values: {latent}")
print(f"compression: {pixels // latent} times")

# 2. Classifier-free guidance on three components
w = 7.5
eps_empty = [0.2, -0.1, 0.0]
eps_text = [0.5, 0.1, -0.2]
pairs = zip(eps_empty, eps_text)
guided = [u + w * (c - u) for u, c in pairs]
values = " ".join(f"{v:.2f}" for v in guided)
print("guided noise:", values)

# 3. Network evaluations for one image
steps = 50
print(f"U-Net passes: {steps * 2}")
````
