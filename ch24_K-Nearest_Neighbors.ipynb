````python
# k nearest neighbours on eight points in the plane
training = [((1, 0), "A"), ((1, 1), "A"), ((1, 2), "A"), ((2, 2), "A"),
            ((5, 5), "B"), ((5, 4), "B"), ((4, 5), "B"), ((2, 3), "B")]
point = (2, 3.5)

def distance(a, b):
    return ((a[0] - b[0]) ** 2 + (a[1] - b[1]) ** 2) ** 0.5

neighbours = sorted(training, key=lambda e: distance(e[0], point))

print(f"neighbours of {point} by distance:")
for x, cls in neighbours[:5]:
    print(f"  {x} {cls} distance {distance(x, point):.3f}")

def classify(k):
    """Majority vote among the k neighbours (odd k, two classes)."""
    votes = [cls for _, cls in neighbours[:k]]
    return max(set(votes), key=votes.count)

for k in (1, 3, 5):
    print(f"k = {k}: class {classify(k)}")
````
