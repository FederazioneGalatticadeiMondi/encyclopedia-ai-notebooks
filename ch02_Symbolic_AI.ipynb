```python
from collections import deque

BLOCKS = ["A", "B", "C"]

def operators(blocks):
    """STRIPS actions 'move x from y to z' (y and z: a block or the table)."""
    places = blocks + ["table"]
    for x in blocks:
        for y in places:
            for z in places:
                if len({x, y, z}) < 3:
                    continue
                pre = {("on", x, y), ("clear", x)}
                add = {("on", x, z)}
                delete = {("on", x, y)}
                if z != "table":
                    pre.add(("clear", z))
                    delete.add(("clear", z))
                if y != "table":
                    add.add(("clear", y))
                yield (f"move {x} from {y} to {z}",
                       frozenset(pre), frozenset(add), frozenset(delete))

def plan(initial, goal, actions):
    """Breadth-first search: returns the plan with the fewest moves."""
    start = frozenset(initial)
    frontier = deque([(start, [])])
    visited = {start}
    while frontier:
        state, steps = frontier.popleft()
        if goal <= state:
            return steps
        for name, pre, add, delete in actions:
            if pre <= state:
                new = (state - delete) | add
                if new not in visited:
                    visited.add(new)
                    frontier.append((new, steps + [name]))
    return None

# Sussman anomaly: C on A; A and B on the table
initial = {("on", "C", "A"), ("on", "A", "table"), ("on", "B", "table"),
           ("clear", "C"), ("clear", "B")}
goal = {("on", "A", "B"), ("on", "B", "C")}

for step in plan(initial, goal, list(operators(BLOCKS))):
    print(step)
```
