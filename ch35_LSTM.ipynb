````python
import math

def sigmoid(z):
    return 1 / (1 + math.exp(-z))

# 1. One LSTM step with a scalar state and assigned pre-activations
c_prev = 1.0
f = sigmoid(2.0)        # forget gate
i = sigmoid(0.0)        # input gate
g = math.tanh(0.0)      # candidate content
o = sigmoid(0.0)        # output gate
c = f * c_prev + i * g  # additive memory update
h = o * math.tanh(c)    # hidden state
print(f"LSTM step: f = {f:.3f}, i = {i:.3f}, o = {o:.3f}, "
      f"c = {c:.3f}, h = {h:.3f}")

# 2. Write 0.8 at the first step and keep it for another 19 steps
def final_memory(forget_gate, steps=20):
    c = 0.0 * forget_gate + 1.0 * 0.8     # first step: write
    for _ in range(steps - 1):
        c = forget_gate * c + 0.0         # input gate closed
    return c

print(f"memory after 20 steps with f = 1.0: {final_memory(1.0):.3f}")
print(f"memory after 20 steps with f = 0.5: {final_memory(0.5):.2e}")
````
