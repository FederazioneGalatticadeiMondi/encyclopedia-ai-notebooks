No installation is needed. The code uses only Python's standard library.

````python
REFERENCE_YEAR = 2026          # year used to compute age

class Frame:
    def __init__(self, name, parent=None, **slots):
        self.name = name
        self.parent = parent       # "is-a" relation
        self.slots = dict(slots)

    def value(self, attribute):
        """Look for the slot in the frame, then climb the hierarchy (inheritance)."""
        frame = self
        while frame is not None:
            if attribute in frame.slots:
                v = frame.slots[attribute]
                if callable(v):            # attached procedure
                    v = v(self)
                return v, frame.name
            frame = frame.parent
        return None, None

animal = Frame("animal", breathes=True)
bird = Frame("bird", animal, flies=True, legs=2)              # defaults
penguin = Frame("penguin", bird, flies=False)                  # exception
tweety = Frame("tweety", bird)
pingu = Frame("pingu", penguin, year_of_birth=2019,
              age=lambda f: REFERENCE_YEAR - f.value("year_of_birth")[0])

for individual, attributes in [(tweety, ["flies", "legs", "breathes"]),
                               (pingu, ["flies", "legs", "breathes", "age"])]:
    for attribute in attributes:
        v, source = individual.value(attribute)
        print(f"{individual.name}.{attribute} = {v}  (from {source})")

# Non-monotonic reasoning: new information overrides the default
tweety.slots["flies"] = False      # Tweety has an injured wing
v, source = tweety.value("flies")
print(f"after the new information: tweety.flies = {v}  (from {source})")
````
