````python
import math

# Relative compute and loss of three small models (example)
compute = [1, 10, 100]
loss = [4.0, 3.2, 2.56]

# Linear regression on a log scale: log L = q + m log C
xs = [math.log10(c) for c in compute]
ys = [math.log10(l) for l in loss]
mx = sum(xs) / len(xs)
my = sum(ys) / len(ys)
num = sum((x - mx) * (y - my) for x, y in zip(xs, ys))
den = sum((x - mx) ** 2 for x in xs)
m = num / den
q = my - m * mx
print(f"exponent b: {-m:.4f}")

# Prediction for larger models
for c in [1000, 10000]:
    predicted = 10 ** (q + m * math.log10(c))
    print(f"compute {c}: predicted loss {predicted:.3f}")
````
