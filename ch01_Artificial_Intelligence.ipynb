```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Data: 150 flowers, 4 measurements each, 3 species
X, y = load_iris(return_X_y=True)

# 2. Split: 75% training, 25% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

# 3-4. Standardisation + model in a single pipeline
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)

# 5. Evaluation on unseen data
predictions = model.predict(X_test)
print(f"Test accuracy: {accuracy_score(y_test, predictions):.2f}")
```
