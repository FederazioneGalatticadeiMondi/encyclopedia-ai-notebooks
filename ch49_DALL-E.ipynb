````python
import math

# 1. Tokens of the first version of DALL·E
img_side, tok_side = 256, 32
img_vocab, max_text = 8192, 256
img_tokens = tok_side * tok_side
print(f"image tokens: {img_tokens}")
print(f"maximum sequence: {img_tokens + max_text}")

# 2. Compression of the discrete autoencoder
pixel_bits = img_side * img_side * 3 * 8
token_bits = img_tokens * int(math.log2(img_vocab))
print(f"bits: {pixel_bits} versus {token_bits}")
print(f"compression: {pixel_bits / token_bits:.1f} times")

# 3. Re-ranking the candidates with CLIP scores
scores = [0.21, 0.35, 0.18, 0.31, 0.27]
indices = range(len(scores))
order = sorted(indices, key=lambda i: -scores[i])
print("best candidates:", order[:2])
````
