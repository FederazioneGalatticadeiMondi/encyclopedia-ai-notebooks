````python
import math
from collections import Counter

# Training messages: spam and legitimate
training = {
    "spam": ["free offer", "free money free"],
    "legitimate": ["meeting tomorrow", "offer meeting"],
}
vocabulary = {word for texts in training.values()
              for t in texts for word in t.split()}

# Training: prior probabilities and word counts per class
total_messages = sum(len(t) for t in training.values())
priors, counts, totals = {}, {}, {}
for cls, texts in training.items():
    priors[cls] = len(texts) / total_messages
    counts[cls] = Counter(w for t in texts for w in t.split())
    totals[cls] = sum(counts[cls].values())

def log_prob(word, cls, alpha=1):
    """Logarithm of P(word | class) with Laplace smoothing."""
    numerator = counts[cls][word] + alpha
    denominator = totals[cls] + alpha * len(vocabulary)
    return math.log(numerator / denominator)

def classify(message):
    scores = {c: math.log(priors[c]) +
                 sum(log_prob(w, c) for w in message.split())
              for c in training}
    top = max(scores.values())
    total = sum(math.exp(s - top) for s in scores.values())
    p_spam = math.exp(scores["spam"] - top) / total
    cls = "spam" if p_spam > 0.5 else "legitimate"
    print(f'"{message}": P(spam) = {p_spam:.3f} -> {cls}')

classify("free offer")
classify("meeting tomorrow offer")
````
