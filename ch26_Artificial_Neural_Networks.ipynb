````python
# Threshold neurons with hand-chosen weights
def neuron(inputs, weights, bias):
    """McCulloch-Pitts neuron: 1 if the weighted sum is above zero."""
    z = sum(x * w for x, w in zip(inputs, weights)) + bias
    return 1 if z > 0 else 0

def xor_network(x1, x2):
    """Hidden layer with OR and AND; output: OR and not AND."""
    h_or = neuron((x1, x2), (1, 1), -0.5)
    h_and = neuron((x1, x2), (1, 1), -1.5)
    return h_or, h_and, neuron((h_or, h_and), (1, -1), -0.5)

print("x1 x2 | OR AND XOR")
for x1 in (0, 1):
    for x2 in (0, 1):
        o, a, x = xor_network(x1, x2)
        print(f" {x1}  {x2} |  {o}   {a}   {x}")

# Number of parameters of a fully connected network
def parameters(layers):
    return sum(n * m + m for n, m in zip(layers, layers[1:]))

print("parameters of the 784-128-10 network:", parameters([784, 128, 10]))
````
