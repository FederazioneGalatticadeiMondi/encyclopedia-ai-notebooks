````python
# Hours of study -> exam outcome (1 = passed); two examples are "noisy"
DATA = [(1, 0), (2, 0), (3, 0), (4, 1), (5, 0),
        (6, 1), (7, 1), (8, 1), (9, 1), (10, 1)]

def train_memoriser(data):
    """Remember the examples; for new cases, give the most frequent outcome."""
    table = dict(data)
    passed = sum(y for _, y in data)
    majority = 1 if passed >= len(data) - passed else 0
    return lambda x: table.get(x, majority)

def train_threshold(data):
    """Choose the threshold t that minimises the errors of the rule 'x >= t'."""
    candidates = sorted(x for x, _ in data) + [max(x for x, _ in data) + 1]
    def errors(t):
        return sum((1 if x >= t else 0) != y for x, y in data)
    t = min(candidates, key=errors)
    model = lambda x: 1 if x >= t else 0
    model.threshold = t
    return model

def accuracy(model, data):
    return sum(model(x) == y for x, y in data) / len(data)

def cross_validation(train, data, k=5):
    """Each example is evaluated by a model that has not seen it."""
    correct = 0
    for fold in range(k):
        test = [d for i, d in enumerate(data) if i % k == fold]
        training = [d for i, d in enumerate(data) if i % k != fold]
        model = train(training)
        correct += sum(model(x) == y for x, y in test)
    return correct / len(data)

print("threshold learned on all the data:", train_threshold(DATA).threshold)
for name, train in [("memoriser", train_memoriser),
                    ("threshold", train_threshold)]:
    model = train(DATA)
    print(f"{name}: training {accuracy(model, DATA):.2f}, "
          f"cross-validation "
          f"{cross_validation(train, DATA):.2f}")
````
