````python
import math

def normalise(v):
    n = math.sqrt(sum(x * x for x in v))
    return [x / n for x in v]

def cosine(a, b):
    a, b = normalise(a), normalise(b)
    return sum(x * y for x, y in zip(a, b))

# Example vectors in the shared image-text space
image = [0.6, 0.8, 0.0]
texts = {
    "a photo of a cat": [0.8, 0.6, 0.0],
    "a photo of a dog": [0.6, 0.0, 0.8],
    "a photo of a car": [0.0, 0.0, 1.0],
}

scale = 10.0  # inverse of the temperature
sim = {t: cosine(image, v) for t, v in texts.items()}
z = sum(math.exp(scale * s) for s in sim.values())
for t, s in sim.items():
    p = math.exp(scale * s) / z
    print(f"{t}: cosine = {s:.2f}, probability = {p:.4f}")
print("prediction:", max(sim, key=sim.get))
````
