````python
import math

# Scaled dot-product attention: 1 query, 3 keys and values
q = [1, 0]
keys = [[1, 0], [0, 1], [1, 1]]
values = [[10, 0], [0, 10], [5, 5]]
d = len(q)

def dot(a, b):
    return sum(x * y for x, y in zip(a, b))

scores = [dot(q, k) / math.sqrt(d) for k in keys]
exponentials = [math.exp(s) for s in scores]
weights = [e / sum(exponentials) for e in exponentials]
output = [sum(w * v[i] for w, v in zip(weights, values)) for i in range(2)]

print("scores:", " ".join(f"{s:.3f}" for s in scores))
print("attention weights:", " ".join(f"{w:.3f}" for w in weights))
print("output:", " ".join(f"{o:.3f}" for o in output))
````
