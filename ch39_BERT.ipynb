````python
# WordPiece tokenisation with longest-match search
vocabulary = {"[UNK]", "the", "play", "##er", "##ing", "foot", "##ballers"}

def wordpiece(word):
    tokens, start = [], 0
    while start < len(word):
        end, found = len(word), None
        while end > start:
            piece = word[start:end]
            if start > 0:
                piece = "##" + piece
            if piece in vocabulary:
                found = piece
                break
            end -= 1
        if found is None:
            return ["[UNK]"]          # no fragment: unknown word
        tokens.append(found)
        start = end
    return tokens

for word in ["the", "player", "playing", "footballers", "racket"]:
    print(word, "->", wordpiece(word))

# An example for the masked language model
sequence = ["[CLS]"] + wordpiece("the") + wordpiece("player") + ["[SEP]"]
masked = list(sequence)
masked[3] = "[MASK]"            # position chosen for the example
print("sequence:", " ".join(sequence))
print("masked:", " ".join(masked))
````
