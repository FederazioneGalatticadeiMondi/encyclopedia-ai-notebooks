````python
# Transaction amount -> label (1 = fraud, 0 = legitimate)
TRAINING = ([(x, 0) for x in [20, 35, 50, 15, 40, 60, 25, 30, 45, 55]]
            + [(480, 1), (520, 1)])
TEST = ([(x, 0) for x in [10, 25, 40, 55, 70, 30, 45, 60, 20, 35,
                          50, 65, 15, 80, 90, 300, 25, 40]]
        + [(450, 1), (200, 1)])

def train_centroids(data):
    """Compute the average amount of each class; predict the nearest class."""
    centroids = {}
    for cls in sorted({y for _, y in data}):
        values = [x for x, y in data if y == cls]
        centroids[cls] = sum(values) / len(values)
    model = lambda x: min(centroids, key=lambda c: abs(x - centroids[c]))
    return model, centroids

def always_legitimate(x):
    return 0

def metrics(model, data):
    tp = sum(1 for x, y in data if model(x) == 1 and y == 1)
    fp = sum(1 for x, y in data if model(x) == 1 and y == 0)
    fn = sum(1 for x, y in data if model(x) == 0 and y == 1)
    tn = len(data) - tp - fp - fn
    accuracy = (tp + tn) / len(data)
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    f1 = (2 * precision * recall / (precision + recall)
          if precision + recall else 0.0)
    return tp, fp, fn, tn, accuracy, precision, recall, f1

model, centroids = train_centroids(TRAINING)
print(f"learned centroids: legitimate {centroids[0]}, fraud {centroids[1]}")
for name, m in [("always legitimate", always_legitimate),
                ("centroids", model)]:
    tp, fp, fn, tn, acc, prec, rec, f1 = metrics(m, TEST)
    print(f"{name}: TP={tp} FP={fp} FN={fn} TN={tn} | accuracy {acc:.2f} "
          f"precision {prec:.2f} recall {rec:.2f} F1 {f1:.2f}")
````
