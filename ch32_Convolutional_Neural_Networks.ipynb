````python
# 2D convolution (cross-correlation) without padding and with stride 1
image = [[0, 0, 0, 1, 1] for _ in range(5)]   # vertical edge
kernel = [[-1, 0, 1],                          # vertical-edge
          [-1, 0, 1],                          # detector
          [-1, 0, 1]]

def convolution(img, k):
    n, m = len(img), len(k)
    side = n - m + 1                           # (n + 2p - k) / s + 1
    return [[sum(k[u][v] * img[i + u][j + v]
                 for u in range(m) for v in range(m))
             for j in range(side)]
            for i in range(side)]

feature_map = convolution(image, kernel)
print(f"map size: {len(feature_map)} x {len(feature_map[0])}")
for row in feature_map:
    print(" ".join(str(v) for v in row))

# Parameters: 3x3 convolution, 3 channels, 64 filters vs. a dense layer
conv = 3 * 3 * 3 * 64 + 64
dense = 224 * 224 * 3 * 64 + 64
print("parameters of the convolutional layer:", conv)
print("parameters of the fully connected layer:", dense)
````
