````python
history = [
    ("system", "You are an assistant answering in English."),
    ("user", "What is a transformer?"),
    ("assistant", "An attention-based architecture."),
    ("user", "Who proposed it?"),
    ("assistant", "Vaswani and colleagues in 2017."),
    ("user", "And what does GPT add?"),
]

def truncate(history, budget):
    """Keeps the system message and the most recent messages within the budget."""
    system, rest = history[0], history[1:]
    used = len(system[1].split())
    kept = []
    for role, text in reversed(rest):
        n = len(text.split())
        if used + n > budget:
            break
        kept.insert(0, (role, text))
        used += n
    return [system] + kept, used

budget = 20
context, used = truncate(history, budget)
for role, text in context:
    print(f"{role}: {text}")
print(f"words used: {used} of {budget}")
print(f"messages excluded: {len(history) - len(context)}")
````
